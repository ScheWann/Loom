# Quality Control and Annotation of 2 um Bin2cell Output

This notebook follows `1.bin2cell`. It takes the aggregated **single-cell** AnnData (`skin_..._b2c.h5ad`), runs quality control (QC), normalization, dimensionality reduction and clustering, annotates cell types with CellTypist, and exports the QC'd data.

**Pipeline:** read -> pre-filter -> compute QC metrics -> flag and remove low-quality cells (MAD-based) -> normalize + highly variable genes + cell-type annotation -> PCA/UMAP/Leiden -> spatial visualization -> export.

```python
import scanpy as sc
import celltypist
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
from scipy.stats import median_abs_deviation
from PIL import Image
Image.MAX_IMAGE_PIXELS = None

import time

notebook_start = time.time()
```

## 1. Load data

Read the single-cell expression matrix produced by the previous bin2cell step. Each obs row is one cell, carrying fields such as `object_id` and `bin_count`.

```python
cdata = sc.read_h5ad('skin_TXK6Z4X_A1_2um_b2c.h5ad')
cdata
```

## 2. Pre-filter and prepare

- Keep only cells built from **at least 8 bins** (`bin_count > 7`), removing tiny, unreliable "cells".
- Round expression values to integers: the `seurat_v3` highly-variable-gene flavor requires integer raw counts.
- Back up the current raw counts in `cdata.raw` so original values can be recovered after normalization and used for the final export.

```python
cdata = cdata[cdata.obs['bin_count']>7] # min 8 bins

# need integers for seuratv3 hvgs
cdata.X.data = np.round(cdata.X.data)
cdata.raw = cdata.copy()
```

## 3. Compute QC metrics

Flag mitochondrial genes (prefix `MT-`) and run `calculate_qc_metrics` to compute per-cell metrics: `total_counts`, `n_genes_by_counts`, `pct_counts_mt`, `pct_counts_in_top_20_genes`, etc. A high mitochondrial fraction usually indicates damaged or dying cells.

```python
cdata.var["mt"] = cdata.var_names.str.startswith("MT-")
sc.pp.calculate_qc_metrics(cdata, qc_vars=["mt"], percent_top=[20], inplace=True)

cdata
```

### Visualize QC metrics

Use violin, distribution, and scatter plots to inspect the distribution of each QC metric and guide the filtering thresholds below.

```python
p1 = sc.pl.violin(cdata, "pct_counts_mt")
p2 = sns.displot(cdata.obs["total_counts"])
p3 = sc.pl.scatter(cdata, "total_counts", "n_genes_by_counts", color="pct_counts_mt")
```

## 4. Flag low-quality cells with MAD

Use robust **median absolute deviation (MAD)** outlier detection instead of hard manual cutoffs: a metric is an outlier if it deviates from the median by more than `nmads` times the MAD.

- Count-based metrics (`log1p_total_counts`, `log1p_n_genes_by_counts`, `pct_counts_in_top_20_genes`) use **5 MADs**.
- Mitochondrial fraction `pct_counts_mt` is stricter: **3 MADs** and additionally must be `> 8%` to count as an outlier.

```python
def is_outlier(adata, metric: str, nmads: int):
    M = adata.obs[metric]
    outlier = (M < np.median(M) - nmads * median_abs_deviation(M)) | (
        np.median(M) + nmads * median_abs_deviation(M) < M
    )
    return outlier
```

```python
cdata.obs["outlier"] = (
    is_outlier(cdata, "log1p_total_counts", 5)
    | is_outlier(cdata, "log1p_n_genes_by_counts", 5)
    | is_outlier(cdata, "pct_counts_in_top_20_genes", 5)
)
cdata.obs.outlier.value_counts()
```

```python
cdata.obs["mt_outlier"] = is_outlier(cdata, "pct_counts_mt", 3) | (
    cdata.obs["pct_counts_mt"] > 8
)
cdata.obs.mt_outlier.value_counts()
```

### Apply filtering

Remove cells flagged as `outlier` or `mt_outlier`, and print the cell count before and after filtering.

```python
print(f"Total number of cells: {cdata.n_obs}")
cdata = cdata[(~cdata.obs.outlier) & (~cdata.obs.mt_outlier)].copy()

print(f"Number of cells after filtering of low quality cells: {cdata.n_obs}")
```

### Re-check after filtering

Re-plot the scatter to confirm low-quality cells are gone and the metric distributions are cleaner.

```python
p1 = sc.pl.scatter(cdata, "total_counts", "n_genes_by_counts", color="pct_counts_mt")
```

## 5. Normalization, highly variable genes, and cell-type annotation

- `normalize_total` + `log1p`: library-size normalization and log transform.
- `highly_variable_genes` (`seurat_v3`, top 5000): select highly variable genes.
- `celltypist.annotate`: predict each cell's type using the adult-skin reference model `Adult_Human_Skin.pkl`, writing `predicted_labels` / `conf_score`.

```python
sc.pp.normalize_total(cdata,target_sum=1e4)
sc.pp.log1p(cdata)
sc.pp.highly_variable_genes(cdata,n_top_genes=5000,flavor="seurat_v3")
b2c_predictions = celltypist.annotate(cdata, model = 'Adult_Human_Skin.pkl')
cdata = b2c_predictions.to_adata()
cdata
```

## 6. Dimensionality reduction

Subset to highly variable genes, then `scale` -> `PCA` -> neighbor graph -> `UMAP` to prepare for clustering and visualization.

```python
cdata = cdata[:, cdata.var["highly_variable"]].copy()
sc.pp.scale(cdata, max_value=10)
sc.pp.pca(cdata, use_highly_variable=True)
sc.pp.neighbors(cdata)
sc.tl.umap(cdata)
```

## 7. Leiden clustering

Run Leiden clustering on the neighbor graph (`resolution=1`) and color the clusters on the UMAP.

```python
sc.set_figure_params(dpi=100,fontsize=10,)

sc.tl.leiden(cdata,resolution=1,key_added='leiden')
sc.pl.umap(cdata,color=['leiden'],size=2,wspace=0.25,frameon=False)
```

## 8. Spatial visualization

### Spatial expression of a single gene
Using the tissue image as background and the `spatial_cropped_150_buffer` coordinates, overlay each cell's `NLGN4Y` expression. `NLGN4Y` is a Y-chromosome gene, useful for inspecting spatial distribution / sample sex signal.

```python
gene = "NLGN4Y"

bg_img = Image.open("../../Example_Data/H1-TXK6Z4X-A1_2um/H1-TXK6Z4X-A1_2um_lite.jpeg")

bg_width, bg_height = bg_img.size

coords = cdata.obsm["spatial_cropped_150_buffer"]
expr = cdata[:, gene].X.toarray().flatten()

plt.figure(figsize=(6, 6))
plt.imshow(bg_img, extent=[0, bg_width, bg_height, 0])
plt.scatter(coords[:, 0], coords[:, 1], c=expr, cmap="viridis", s=0.1, alpha=0.7)
plt.colorbar(label=f"{gene} expression")
plt.title(f"Spatial expression of {gene}")
plt.axis("equal")
plt.xlim(0, bg_width)
plt.ylim(bg_height, 0) 
plt.show()
```

### Spatial distribution of predicted cell types

Using the `spatial` coordinates, plot the CellTypist-predicted cell types colored on tissue space to examine the spatial organization of each cell type.

```python
coords = cdata.obsm["spatial"]
labels = cdata.obs["predicted_labels"]

df = pd.DataFrame({
    "x": coords[:, 0],
    "y": coords[:, 1],
    "cell_type": labels
})

plt.figure(figsize=(8, 8))
sns.scatterplot(data=df, x="x", y="y", hue="cell_type", s=5, alpha=0.8, linewidth=0)
plt.gca().invert_yaxis()
plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left")
plt.title("Spatial distribution of predicted cell types")
plt.axis("equal")
plt.show()
```
